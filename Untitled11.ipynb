```python
import numpy as np
import scipy.stats as stats

# 1. Menyiapkan Data Sebanyak 100 Sampel
# Menggunakan data random dengan distribusi normal (misal: mean=50, std=10)
# Anda bisa mengganti bagian ini dengan data riil milik Anda sendiri jika ada.
np.random.seed(42)  # Agar hasil data random tetap konsisten saat dijalankan
data = np.random.normal(loc=50, scale=10, size=100)

print(f"Data sampel (10 data pertama): {data[:10]}...\n")

# ==========================================
# 2. Point a: Estimasi Titik dan Interval (Distribusi Z)
# Tingkat kesalahan 5% (alpha = 0.05), berarti tingkat keyakinan (confidence level) = 95%
# ==========================================
alpha = 0.05
n = len(data)

# Estimasi Titik (Rata-rata sampel)
point_estimate = np.mean(data)

# Standar deviasi dari data sampel
std_dev = np.std(data, ddof=1)

# Nilai Z untuk tingkat keyakinan 95% (dua sisi)
z_score = stats.norm.ppf(1 - alpha / 2)

# Margin of Error
margin_of_error = z_score * (std_dev / np.sqrt(n))

# Batas Bawah dan Batas Atas Interval Estimasi
lower_bound = point_estimate - margin_of_error
upper_bound = point_estimate + margin_of_error

print("--- JAWABAN BAGIAN A ---")
print(f"Estimasi Titik (Rata-rata): {point_estimate:.4f}")
print(f"Nilai Z (untuk alpha 5%): {z_score:.4f}")
print(
    f"Interval Estimasi 95%: [{lower_bound:.4f}, {upper_bound:.4f}]\n"
)

# ==========================================
# 3. Point b: Resampling (Jackknife & Bootstrapping)
# ==========================================

# --- Metode Jackknife ---
# Jackknife mengeluarkan 1 data secara bergantian sebanyak n kali
jackknife_means = []
for i in range(n):
  # Membuat data baru tanpa elemen ke-i
  resample_data = np.delete(data, i)
  jackknife_means.append(np.mean(resample_data))

mean_jackknife = np.mean(jackknife_means)

print("--- JAWABAN BAGIAN B ---")
print(f"Rata-rata dari hasil resampling Jackknife: {mean_jackknife:.4f}")

# --- Metode Bootstrapping ---
# Bootstrapping mengambil sampel dengan pengembalian (with replacement) sebanyak 1000 iterasi
n_iterations = 1000
bootstrap_means = []

for _ in range(n_iterations):
  # Mengambil sampel acak dari data dengan ukuran yang sama (100) secara 'with replacement'
  resample_bootstrap = np.random.choice(data, size=n, replace=True)
  bootstrap_means.append(np.mean(resample_bootstrap))

mean_bootstrap = np.mean(bootstrap_means)

print(f"Jumlah Iterasi Bootstrapping: {n_iterations}")
print(f"Rata-rata dari hasil resampling Bootstrapping: {mean_bootstrap:.4f}")
```
